```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd


VP = 50
VN = 40
FP = 10
FN = 5

N = VP + VN + FP + FN


print("Matriz de Confusão:")
print(f"VP: {VP}, FN: {FN}")
print(f"FP: {FP}, VN: {VN}\n")


conf_matrix = pd.DataFrame([[VP, FN], [FP, VN]],index=["Positivo (Real)", "Negativo (Real)"],columns=["Positivo (Previsto)", "Negativo (Previsto)"])

sensibilidade = VP / (VP + FN)  # Recall
especificidade = VN / (FP + VN)
acuracia = (VP + VN) / N
precisao = VP / (VP + FP)
f_score = 2 * (precisao * sensibilidade) / (precisao + sensibilidade)


print("Resultados das métricas:")
print(f"Sensibilidade (Recall): {sensibilidade:.3f}")
print(f"Especificidade: {especificidade:.3f}")
print(f"Acurácia: {acuracia:.3f}")
print(f"Precisão: {precisao:.3f}")
print(f"F-Score: {f_score:.3f}\n")


plt.figure(figsize=(6,4))
sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues")
plt.title("Matriz de Confusão")
plt.show()
```
